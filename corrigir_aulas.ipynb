# Correção de aulas

Este notebook ajuda a corrigir caracteres estranhos e revisar a ortografia nas páginas de aula do projeto.

````python
## 1. Importar bibliotecas

```python
from pathlib import Path
import re
from typing import List
```
````

````python
## 2. Carregar arquivos das aulas

```python
root = Path('aulas')
files = sorted(root.rglob('*.html'))
files[:5]
```
````

````python
## 3. Identificar problemas de encoding e caracteres estranhos

```python
bad_markers = ['Ã', 'â', 'Â', 'Å', 'â€', 'â€™', 'â€œ', 'â€']

for path in files:
    text = path.read_text(encoding='utf-8', errors='ignore')
    if any(marker in text for marker in bad_markers):
        print(path.name)
```
````

````python
## 4. Corrigir caracteres estranhos no texto

```python
from pathlib import Path


def fix_mojibake(text: str) -> str:
    if not text:
        return text

    current = text
    for _ in range(5):
        try:
            candidate = current.encode('latin-1').decode('utf-8')
        except Exception:
            break
        if candidate == current:
            break
        current = candidate

    return current


root = Path('aulas')
for path in sorted(root.rglob('*.html')):
    original = path.read_text(encoding='utf-8', errors='ignore')
    fixed = fix_mojibake(original)
    if fixed != original:
        path.write_text(fixed, encoding='utf-8')
        print(f'Corrigido: {path}')
```
````

````python
## 5. Revisar e corrigir ortografia nas aulas

```python
# Ajustes mínimos de ortografia para palavras comuns que podem aparecer nas aulas.
replacements = {
    'apresentaÃ§Ã£o': 'apresentação',
    'apresentaçãƒÂ£o': 'apresentação',
    'apresentaÃ§Ãµes': 'apresentações',
    'apresentaÃ§Ã£o': 'apresentação',
    'cÃ³digo': 'código',
    'sÃ£o': 'são',
    'nÃ£o': 'não',
    'fÃ³rmula': 'fórmula',
    'dÃ©cadas': 'décadas',
    'informaÃ§Ã£o': 'informação',
    'tÃtulo': 'título',
    'pÃ¡gina': 'página',
    'conhecÃªncia': 'conhecimento',
    'quase sempre': 'quase sempre',
}

for path in files:
    text = path.read_text(encoding='utf-8', errors='ignore')
    updated = text
    for old, new in replacements.items():
        updated = updated.replace(old, new)
    if updated != text:
        path.write_text(updated, encoding='utf-8')
        print(f'Ortografia ajustada: {path.name}')
```
````
